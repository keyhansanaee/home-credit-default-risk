**2.3 Missing Values Analysis**

```python
missing = pd.DataFrame({
    'Missing Count' : application_train.isna().sum() , 
    'Missing Percent' : (application_train.isna().mean() * 100).round(2)
})

missing = missing.sort_values(
    by = 'Missing Percent' , 
    ascending= False
)

missing
```

```python
extra_missing = missing[missing['Missing Percent'] > 50]
extra_missing
```

**2.9 Constant Features**

```python
constant_columns = []
for col in application_train.columns :
    if application_train[col].nunique() == 1:
        constant_columns.append(col)
constant_columns
```
